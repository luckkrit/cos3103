```python
from pymongo import MongoClient
uri = "mongodb://localhost:27017/"

def execute(callable):
    try:
        client = MongoClient(uri)
        callable(client)

        client.close()
    except Exception as e:
        raise Exception("Unable to find the document due to the following error: ", e)

def find(collection,filter={},projection={},skip=0,limit=0):
    for c in collection.find(filter,projection,skip=skip,limit=limit):
        print(c)

def find_one(collection,filter={},projection={},skip=0,limit=0):
    c = collection.find_one(filter,projection,skip=skip,limit=limit)
    print(c)

def insert_one(collection, data):
    collection.insert_one(data)

def insert_many(collection, data):
    collection.insert_many(data)
```
